```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import warnings
warnings.filterwarnings('ignore')

print(" Bibliothèques importées")
```

```python
df = pd.read_csv('../data/processed/dataset_clean.csv')

# Prendre seulement les colonnes numériques (ignorer les labels)
features = ['Poids', 'Volume', 'Conductivite', 'Opacite', 'Rigidite']
X = df[features].dropna()

print(f"Shape : {X.shape}")
print("Labels ignorés — mode non supervisé")
```

```python
inerties = []
K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inerties.append(kmeans.inertia_)

plt.figure(figsize=(10, 5))
plt.plot(K, inerties, 'bo-', markersize=8)
plt.xlabel('Nombre de clusters (k)')
plt.ylabel('Inertie')
plt.title('Méthode du Coude — Choix optimal de k')
plt.xticks(K)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
silhouettes = []

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)
    silhouettes.append(score)
    print(f"k={k} → Silhouette: {score:.4f}")

plt.figure(figsize=(10, 5))
plt.plot(K, silhouettes, 'ro-', markersize=8)
plt.xlabel('Nombre de clusters (k)')
plt.ylabel('Score Silhouette')
plt.title('Score Silhouette par nombre de clusters')
plt.xticks(K)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# k=4 choisi après analyse : meilleur score Silhouette observé à k=4
kmeans_final = KMeans(n_clusters=4, random_state=42, n_init=10)
labels_clusters = kmeans_final.fit_predict(X)

print(f" KMeans appliqué avec k=4")
print(f"Distribution des clusters :")
unique, counts = np.unique(labels_clusters, return_counts=True)
for u, c in zip(unique, counts):
    print(f"   Cluster {u} : {c} points")
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(10, 7))
colors = ['blue', 'red', 'green', 'orange']
for i in range(4):
    mask = labels_clusters == i
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1],
                c=colors[i], label=f'Cluster {i}',
                alpha=0.6, s=30)

plt.title(f'Clustering KMeans (k=4) — Visualisation PCA 2D\nVariance expliquée: {sum(pca.explained_variance_ratio_)*100:.1f}%')
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# === INTERPRETATION DES CLUSTERS ===
X_interpret = X.copy()
X_interpret['Cluster'] = labels_clusters

print("=== PROFIL MOYEN PAR CLUSTER ===")
profil = X_interpret.groupby('Cluster')[features].mean().round(3)
print(profil)

# Heatmap des profils
plt.figure(figsize=(10, 4))
sns.heatmap(profil, annot=True, fmt='.2f', cmap='YlOrRd')
plt.title('Profil moyen des features par cluster')
plt.tight_layout()
plt.show()

# Interprétation textuelle basée sur les valeurs observées
print("\n=== INTERPRETATION ===")
for cluster_id in range(4):
    row = profil.loc[cluster_id]
    print(f"\nCluster {cluster_id} :")
    print(f"  Poids={row['Poids']:.2f} | Volume={row['Volume']:.2f} | "
          f"Conductivite={row['Conductivite']:.2f} | "
          f"Opacite={row['Opacite']:.2f} | Rigidite={row['Rigidite']:.2f}")
```

```python
# === SAUVEGARDE ===
import joblib, os

os.makedirs('../models', exist_ok=True)
os.makedirs('../data/processed', exist_ok=True)

# Sauvegarder le modèle KMeans et le PCA
joblib.dump(kmeans_final, '../models/kmeans_best.pkl')
joblib.dump(pca,          '../models/pca_2d.pkl')

# Sauvegarder le dataset avec les labels de clusters
X_with_labels = X.copy()
X_with_labels['Cluster'] = labels_clusters
X_with_labels.to_csv('../data/processed/dataset_clusters.csv', index=False)

print(" KMeans sauvegardé   → models/kmeans_best.pkl")
print(" PCA sauvegardée     → models/pca_2d.pkl")
print(" Dataset + clusters  → data/processed/dataset_clusters.csv")
```
